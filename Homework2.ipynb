```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[1,2],[2,4]])

def rayleigh_quotient(A, x1, x2):
    numerator =  A[0, 0]*x1**2 + (A[0, 1] + A[1, 0])*x1*x2 + A[1, 1]*x2**2
    denominator = x1**2 + x2**2
    return numerator/denominator
```

```python
x1 = np.linspace(-3,3,100)
x2 = np.linspace(-3,3,100)

X1, X2 = np.meshgrid(x1, x2)

denominator = X1**2 + X2**2

R = np.full_like(X1, np.nan)

mask = denominator > 0


R[mask] = rayleigh_quotient(A, X1[mask], X2[mask])

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(X1, X2, R)

ax.set_xlabel(r"$x_1$")
ax.set_ylabel(r"$x_2$")
ax.set_zlabel(r"$R_A(x)$")
ax.set_title("Rayleigh Quotient")

plt.show()
```
