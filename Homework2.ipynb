The problem asks us to take
$$A=\begin{bmatrix}1 & 2 \\ 2 & 4 \end{bmatrix}$$
and visualize the Rayleigh Quotient $R_A(x)$ for $x \in [-3,3]^2-\{0\}$ with a contour plot.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[1,2],[2,4]])

def rayleigh_quotient(A, x1, x2):
    numerator =  A[0, 0]*x1**2 + (A[0, 1] + A[1, 0])*x1*x2 + A[1, 1]*x2**2
    denominator = x1**2 + x2**2
    return numerator/denominator
```

```python
x1 = np.linspace(-3,3,100)
x2 = np.linspace(-3,3,100)

X1, X2 = np.meshgrid(x1, x2)

denominator = X1**2 + X2**2

R = np.full_like(X1, np.nan)

mask = denominator > 0


R[mask] = rayleigh_quotient(A, X1[mask], X2[mask])

plt.figure(figsize=(8, 7))

contour = plt.contourf(X1, X2, R, levels=30)
plt.colorbar(contour, label=r"$R_A(x)$")

plt.xlabel(r"$x_1$")
plt.ylabel(r"$x_2$")
plt.title("P1. Contour Plot of the Rayleigh Quotient")

plt.axis("equal")
plt.show()
```
